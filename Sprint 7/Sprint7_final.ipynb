**Calculadora de l'índex de massa corporal**

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades.

La funció ha de classificar el resultat en les seves respectives categories

```python
# Creo la función, a la que le paso dos argumentos: peso y altura

def calculadora_IMC(altura, peso):

    # Creo una variable que hace el cálculo del IMC (en la definición pongo la fórmula, uso el peso y la altura que he pedido)
    
    IMC = round(peso/(altura*altura), 2)
    
    #Creo un condicional que clasifique el IMC según el peso y la altura introducidos

    if IMC <= 18.5: 
        return(f"El usuario tiene un IMC de {IMC}, lo que se considera bajo peso.") #creo un retorno que me da la información de la clasificación.
    elif IMC <= 24.9:
        return(f"El usuario tiene un IMC de {IMC}, lo que se considera un peso normal.")
    elif IMC < 29.9:
        return(f"El usuario tiene un IMC de {IMC}, lo que se considera sobrepeso.")
    else:
        return(f"El usuario tiene un IMC de {IMC}, lo que se considera obesidad.")
```

```python
# Pido la información que luego le paso a la función al llamarla

altura = float(input("Altura del usuario en metros."))
peso = float(input("Peso del usuario."))

calculadora_IMC(altura, peso)
```

 **Convertidor de temperatures.**

 Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C),
 Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors,
 de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions

```python
# Creo la función, a la que le paso dos argumentos: valor (número) y unidad de peso

def pedir_temperatura(valor, unidad):    
    # Ahora guardo los datos dentro de un diccionario que creo
    datos_temperatura = {"valor": valor, "unidad": unidad}
    return datos_temperatura
```

```python
# Ahora creo otra función a la que le paso un argumento, que será el diccionario obtenido de la anterior función
def conversion(temperatura):
    #Ahora tomo lo que hay en la clave "valor" del diccionario temperatura (o el que se use en su lugar) y lo guardo en la variable valor y lo mismo con "unidad"
    valor = temperatura["valor"]
    unidad = temperatura["unidad"]
    # Creo una condicional según la cual si la unidad obtenida es "C", entonces se lanza el diccionario celsius y se hace la conversión a fahr y kel
    if unidad == "C":
        dict_celsius ={ #creo un diccionario para celsius
            'cel-fahr': round((valor * 9/5) + 32, 2), #creo la clave de celsius a fahrenheit y en el valor pongo el cálculo
            'cel-kel': round(valor + 273.15, 2) #creo la clave de fahrenheit a celsius y en el valor pongo el cálculo
        }
        return(f"Has introducido {valor} ºC, que son {dict_celsius['cel-fahr']} ºF y {dict_celsius['cel-kel']} K.") #devuelvo el resultado
    # Creo una condicional según la cual si la unidad obtenida es "F", entonces se lanza el diccionario fahrenheit y se hace la conversión a cel y kel
    elif unidad == "F": 
        dict_fahrenheit ={
            'fahr-cel': round((valor - 32) * 5/9, 2), # igual que en la opción de arriba
            'fahr-kel': round((valor - 32) * 5/9 + 273.15, 2)}

        return(f"Has introducido {valor} ºF, que son {dict_fahrenheit['fahr-cel']} ºC y {dict_fahrenheit['fahr-kel']} K.")

    elif unidad == "K": #igual que arriba
        dict_kelvin ={
            'kel-cel': round(valor - 273.15, 2),
            'kel-fahr': round((valor - 273.15) * 9/5 + 32, 2)}

        return(f"Has introducido {valor} K, que son {dict_kelvin['kel-cel']} ºC y {dict_kelvin['kel-fahr']} ºF.")
```

```python
# Pido al usuario los datos que necesito, el valor y la unidad
valor = float(input("¿Qué temperatura deseas convertir? En número.")) #uso float para asegurarme de que sea un valor decimal
unidad = input("¿Qué unidad es? C para Celsius, F para Fahrenheit o K para Kelvin:").upper() #uso upper para asegurarme de que paso el valor en mayúscula

pedir_temperatura(valor, unidad)

# Creo una variable para que contenga el diccionario que devuelve la primera función que he creado y esa variable es la que le paso a la función cuando la llama
diccionario_temperatura = pedir_temperatura(valor, unidad)
conversion(diccionario_temperatura)
```

**Comptador de paraules d'un text.**

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula

```python
# Creo una función para contar las palabras y hago que sea necesario pasarle un texto para que lo cuente

def contador_palabras(texto_a_contar):

    #Creo un diccionario vacío que servirá como contador (en la clave la palabra y en el valor el número de veces que aparece)
    
    ContadorPal = {}
    
    # Creo un bucle que vaya palabra por palabra con .split()
    for palabra in texto_a_contar.split():
        # Hago una limpieza de las palabras quitándoles cualquier símbolo y poniéndolas en minúsculas
        palabra = palabra.lower().strip(".,()¿?¡!;:")
        # Creo un condicional para que añada al diccionario todas las palabras que no están en el contador
        if palabra not in ContadorPal:
            ContadorPal[palabra] = 1
        else:
        # Si la palabra ya estaba en el diccionario, se suma un número a su valor en el diccionario
            ContadorPal[palabra] += 1 
    
    return(ContadorPal)
```

```python
# Pido el texto que luego uso cuando llamo a la función
texto = input("Escribe el texto a analizar.")

contador_palabras(texto)

#Texto ejemplo para utilizar: El Vincent es mi perro y es el perro más bonito sobre cualquier otro perro del mundo incluido mi perro que es el Vincent
```

 **Diccionari invers.**


 Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.


reverse_dictionary({"a": 1, "b": 2, "c": 3})

{1: 'a', 2: 'b', 3: 'c'}

reverse_dictionary({'x': 'apple', 'y': 'banana', 'z': 'banana'})

Error: multiple keys for one value

```python
# Creo una función para invertir un diccionario y hago que sea necesario pasarle un diccionario

def invertir_dic(diccionario_a_invertir):
    # Creo un diccionario vacío en el que se introducirá el diccionario invertido
    dic_invertido = {}
    # Creo un bucle que accede a cada clave y valor del diccionario que queremos invertir
    for clave, valor in diccionario_a_invertir.items():
        # Le pido al bucle que, si el valor ya estaba en el diccionario invertido (el nuevo), muestre el mensaje de error
        if valor in dic_invertido:
            return "Error: varias claves con un mismo valor."
        # Si no, que lo añada de forma normal
        dic_invertido[valor] = clave
    # Y que retorne el nuevo diccionario
    return dic_invertido
```

```python
# Le paso el diccionario a la función y la llamo con él
reverse_dictionary = {"a": 1, "b": 2, "c": 3}

invertir_dic(reverse_dictionary)
```

```python
# Le paso el diccionario a la función y la llamo con él, pero en este caso tengo un valor repetido para comprobar que salta el error
reverse_dictionary2 = {"x": "apple", "y": "banana", "z": "banana"}

invertir_dic(reverse_dictionary2)
```

**Diccionari invers amb duplicats**

Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

reverse_dictionary_plus({'x': 'apple', 'y': 'banana', 'z': 'banana'})
{'apple': 'x', 'banana': ['y', 'z']}

```python
# Creo una función para invertir un diccionario y hago que sea necesario pasarle un diccionario

def invertir_dicconlista(diccionario_invertido):
    # Creo un diccionario vacío en el que se introducirá el diccionario invertido
    dic_invertido2 = {}
    # Creo un bucle que accede a cada clave y valor del diccionario que queremos invertir
    for c, v in diccionario_invertido.items():
        # Le pido al bucle que, si un determinado valor ya tiene una lista de claves, agregue la nueva clave a la lista
        if v in dic_invertido2:
            if isinstance(dic_invertido2[v], list):
                dic_invertido2[v].append(c)
            # Si es una sola clave, la convierte en lista
            else:
                dic_invertido2[v] = [dic_invertido2[v], c]
        #Si ese valor no estaba todavía, se agrega tal cual
        else:
            dic_invertido2[v] = c
            
    return(dic_invertido2)
```

```python
# Le paso el diccionario a la función y la llamo con él. Añado varios valores más con la misma clave para comprobar que funciona

reverse_dictionary_plus = ({'x': 'apple', 'y': 'banana', 'z': 'banana', 'a': 'banana', 'b': 'banana'})

invertir_dicconlista(reverse_dictionary_plus)
```

**Conversió de tipus de dades**

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

conversion([ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']])
([1.3, 1000000000.0, 2.0, 1.0, 1.4, 1.0, 2.0, 3.0, 3.4],
['one, 'seven', '3-1/2', 'not-a-number])

```python
# Creo una función para separar listas y hago que sea necesario pasarle una lista

def separar_floats_nofloats(lista):
    # Creo dos listas vacías para que pueda dividir los datos de la lista que le pasamos
    floats = []
    lista_otros = []
    # Creo un bucle que itere por cada dato de la lista
    for dato in lista:
        # Le digo que si el dato es una lista es una lista o una tupla, cree un nuevo bucle
        if isinstance(dato, (list, tuple)):
            for x in dato: #ya dentro de la lista y la tupla de la lista inicial, pedimos que, si lo que se encuentra es un float (que hayamos podido convertir o que ya existía), lo añada a la lista floats
                try: 
                    floats.append(float(x))
                except: # si no son floats, se añaden a la lista_otros
                    lista_otros.append(x)  
        else: # para los demás datos de la lista que estaban fuera de la lista y la tupla internas, hacemos el mismo proceso:
            try:
                floats.append(float(dato)) #si hay floats, convertidos o ya existentes, se añaden a la lista "floats"
            except:
                lista_otros.append(dato)  #y si hay datos que no son floats, se añaden a la "lista_otros"
                
    return floats, lista_otros
```

```python
# Le paso el diccionario a la función y la llamo con él

lista_inicial = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

separar_floats_nofloats(lista_inicial)
```

**Comptador i endreçador de paraules d'un text.**

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:

```python
# Creo una función para contar y ordenar palabras de un texto y hago que sea necesario pasarle un texto

def contar_ordenar(texto):

    # Dentro del texto, hago un split que me hace una separación por palabras
    palabras = texto.split()
    
    # Creo un diccionario para introducir el conteo de palabras ordenadas
    ContarPal = {}
    
    # Creo un bucle
    for palabra in palabras:
        # Hago una limpieza de las palabras quitándoles cualquier símbolo y poniéndolas en minúsculas
        palabra = palabra.lower().strip(".,;¡!():") 
        # Creo un condicional para que añada al diccionario todas las palabras que no están en el contador
        if palabra not in ContarPal:
            ContarPal[palabra] = 1
        # Si la palabra ya estaba en el diccionario, se suma un número a su valor en el diccionario
        else:
            ContarPal[palabra] += 1 

    # Ordeno el diccionario
    clasificadas_ordenadas = dict(sorted(ContarPal.items()))

    # Creo un nuevo diccionario para colocar el diccionario con la clasificación por letras del alfabeto
    diccionario_mequieres = {}
    # Creo un bucle para poder iterar por todas las palabras
    for palabra, cantidad in clasificadas_ordenadas.items():
        # quito un espacio que no había podido eliminar anteriormente diciéndole que haga este proceso solo si son palabras
        if palabra:
            #cojo la primera letra de la palabra y la pongo en minúscula y la uso como clave si no estaba anteriormente
            first_letter = palabra[0].lower() 
            if first_letter not in diccionario_mequieres:
                diccionario_mequieres[first_letter] = {}
            #añado el resto de palabras y su cantidad
            diccionario_mequieres[first_letter][palabra] = cantidad

    return(diccionario_mequieres)
```

```python
# Le paso un texto en txt a la función, usando el encoding utf-8 para que me acepte las palabras acentuadas (es el encoding del documento)

with open("tu_me_quieres_blanca.txt", encoding="utf-8") as mequieres:
    texto = mequieres.read()

contar_ordenar(texto)
```
